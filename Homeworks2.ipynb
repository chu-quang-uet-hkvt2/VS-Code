```python
print("\n--- Bài 2: Hình Tròn ---")
r = float(input("Nhập bán kính hình tròn: "))

chu_vi_tron = 2 * math.pi * r
dien_tich_tron = math.pi * (r ** 2)

print(f"Chu vi hình tròn: {chu_vi_tron:.2f}")
print(f"Diện tích hình tròn: {dien_tich_tron:.2f}")
```

```python
a = float(input("Nhập cạnh a: "))
b = float(input("Nhập cạnh b: "))
c = float(input("Nhập cạnh c: "))

if a + b > c and a + c > b and b + c > a:
    chu_vi = a + b + c
    
    p = chu_vi / 2
    dien_tich = math.sqrt(p * (p - a) * (p - b) * (p - c))
    
    loai_tam_giac = "Tam giác thường"
    if a == b == c:
        loai_tam_giac = "Tam giác đều"
    elif a == b or a == c or b == c:
        if (abs(a**2 + b**2 - c**2) < 1e-9 or 
            abs(a**2 + c**2 - b**2) < 1e-9 or 
            abs(b**2 + c**2 - a**2) < 1e-9):
             loai_tam_giac = "Tam giác vuông cân"
        else:
            loai_tam_giac = "Tam giác cân"
    elif (abs(a**2 + b**2 - c**2) < 1e-9 or 
          abs(a**2 + c**2 - b**2) < 1e-9 or 
          abs(b**2 + c**2 - a**2) < 1e-9):
        loai_tam_giac = "Tam giác vuông"
        
    print(f"Đây là: {loai_tam_giac}")
    print(f"Chu vi: {chu_vi}")
    print(f"Diện tích: {dien_tich:.2f}")
else:
    print("Khong la ba canh cua mot tam giac")
```

```python
print("\n--- Bài 5: Phương Trình Bậc 2 ---")
a = float(input("Nhập a: "))
b = float(input("Nhập b: "))
c = float(input("Nhập c: "))

if a == 0:
    # Trở về phương trình bậc 1: bx + c = 0
    if b == 0:
        print("Vô số nghiệm" if c == 0 else "Vô nghiệm")
    else:
        print(f"Nghiệm x = {-c / b}")
else:
    delta = b**2 - 4*a*c
    if delta < 0:
        print("Phương trình vô nghiệm thực.")
    elif delta == 0:
        x = -b / (2*a)
        print(f"Phương trình có nghiệm kép x = {x}")
    else:
        x1 = (-b + math.sqrt(delta)) / (2*a)
        x2 = (-b - math.sqrt(delta)) / (2*a)
        print(f"Phương trình có 2 nghiệm phân biệt: x1 = {x1}, x2 = {x2}")
```

```python
print("--- Bài 8: Giải hệ phương trình tuyến tính ---")
a = float(input("Nhập a: "))
b = float(input("Nhập b: "))
m = float(input("Nhập m: ")) 
c = float(input("Nhập c: "))
d = float(input("Nhập d: "))
n = float(input("Nhập n: "))  

D = a * d - c * b
Dx = m * d - n * b
Dy = a * n - c * m

if D != 0:
    x = Dx / D
    y = Dy / D
    print(f"Hệ phương trình có nghiệm duy nhất: (x, y) = ({x}, {y})")
else:
    if Dx != 0 or Dy != 0:
        print("Hệ phương trình vô nghiệm.")
    else:
        print("Hệ phương trình có vô số nghiệm.")
```

```python
print("\n--- Bài 10: Kiểm tra điểm và đường tròn ---")
tam_x = float(input("Nhập hoành độ tâm (Ix): "))
tam_y = float(input("Nhập tung độ tâm (Iy): "))
r = float(input("Nhập bán kính R: "))

a_x = float(input("Nhập hoành độ điểm A (x): "))
a_y = float(input("Nhập tung độ điểm A (y): "))

khoang_cach = math.sqrt((a_x - tam_x)**2 + (a_y - tam_y)**2)

epsilon = 0.000000001

if abs(khoang_cach - r) < epsilon:
    print(f"Điểm A({a_x}, {a_y}) THUỘC đường tròn.")
elif khoang_cach < r:
    print(f"Điểm A({a_x}, {a_y}) nằm TRONG đường tròn (không thuộc đường viền).")
else:
    print(f"Điểm A({a_x}, {a_y}) nằm NGOÀI đường tròn.")
```

```python
print("\n--- Bài 11: Tính x lũy thừa y ---")
x = float(input("Nhập cơ số x: "))
y = float(input("Nhập số mũ y: "))

print(f"Kết quả của {x}^{y} là:")

kq1 = x ** y
print(f"Cách 1 (Toán tử **): {kq1}")

kq2 = math.pow(x, y)
print(f"Cách 2 (math.pow): {kq2}")

kq3 = pow(x, y)
print(f"Cách 3 (built-in pow): {kq3}")

if x > 0:
    kq4 = math.exp(y * math.log(x))
    print(f"Cách 4 (Logarit & Exp): {kq4}")
else:
    print("Cách 4 không áp dụng được vì cơ số x <= 0")
```
